```python
import pandas as pd

# Load data - skip the first row of headers if Bloomberg added metadata
df = pd.read_csv('NVDA.csv', skiprows=1)

# Bloomberg has exported data as: [Date, NVDA, Sentiment, Date.1, TSM, Date.2, ASML]
# We only want one Date column and the data columns
# Adjust the numbers below [0, 1, 2, 4, 6] based on your actual CSV columns
df_clean = df.iloc[:, [0, 1, 2, 4, 6]].copy()
df_clean.columns = ['Date', 'NVDA', 'Sentiment', 'TSM', 'ASML']

# Convert Date to actual time objects and set as index- This helps slicing and filtering, eases accessing time based properties..

df_clean['Date'] = pd.to_datetime(df_clean['Date'])
df_clean = df_clean.sort_values('Date').set_index('Date')

# Fill "Holes" (if one market was closed while another was open)
df_clean = df_clean.ffill().dropna()
```

```python
# @title {"display-mode":"form"}
from IPython.display import HTML

display(HTML("""

  <h1> Transform to Returns & Create "Lags" </h1>
<br>
<p1>
  A Stationary series is one where the statistical properties—like the mean (average) and variance (volatility) don't change over time.
  Stock prices are Non-Stationary. They have "unit roots," meaning they wander off and don't return to a fixed average.
</p1>
  <br>
  <br>
<p1>
  In a Spurious Regression, the math gives you a high R^2 (Coefficient of Determination) and a significant p-value. We could regress NVIDIA's stock price against The Number of People Wearing Sneakers in New York, and if both have increased over the last 10 years, the math will say they are 99% correlated. The problem with it is we aren't measuring how NVDA reacts to sneakers; you are just measuring the passage of time

</p1>


"""))
```

```python
# 1. Calculate Daily % Change
returns = df_clean[['NVDA', 'TSM', 'ASML']].pct_change()

# 2. Create the LAG: Yesterday's Supplier move to predict Today's NVDA
returns['TSM_Lag1'] = returns['TSM'].shift(1)
returns['ASML_Lag1'] = returns['ASML'].shift(1)

# 3. Add back the Sentiment score
returns['Sentiment'] = df_clean['Sentiment']

# Drop the first row (which becomes NaN due to the shift/pct_change)
data_final = returns.dropna()
```

```python
# @title {"display-mode":"form"}
from IPython.display import HTML

display(HTML("""

  <h1> OLS Regression </h1>



"""))
```

```python
import statsmodels.api as sm

# Define Y (Target: NVDA Return) and X (Predictors: Supplier Lags + Sentiment)
Y = data_final['NVDA']
X = data_final[['TSM_Lag1', 'ASML_Lag1', 'Sentiment']]
X = sm.add_constant(X) # Adds the intercept term

# Run the model
model = sm.OLS(Y, X).fit()

# PRINT THE RESULTS
print(model.summary())

```
